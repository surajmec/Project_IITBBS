```python
import pandas as pd
import json
```

```python
df = pd.read_csv("dataset/oct_reports.csv")

print(df.shape)

df.head()
```

```python
instruction = """
You are an ophthalmology AI assistant.

Read the OCT clinical report carefully.

Predict whether the patient belongs to

Healthy

or

Schizophrenia.

Respond using ONLY one word.

Healthy

or

Schizophrenia.
"""
```

```python
dataset = []

for _, row in df.iterrows():

    sample = {

        "instruction": instruction,

        "input": row["report"],

        "output": row["label"]

    }

    dataset.append(sample)
```

```python
dataset[0]
```

```python
with open("dataset/oct_instruction.jsonl","w",encoding="utf-8") as f:

    for sample in dataset:

        f.write(json.dumps(sample))

        f.write("\n")
```
